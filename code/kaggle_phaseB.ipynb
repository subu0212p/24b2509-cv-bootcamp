# Inter-IIT CV bootcamp – Phase B
Task 5 best method (+ other methods on synthetic labels for Q5), then builds the submission files.
Add the **Phase A notebook output** as an input. Accelerator: **GPU T4 x2**.

```python
import glob, os, shutil, subprocess, sys
inp = "/kaggle/input"
data_dir = os.path.dirname(sorted(glob.glob(inp + "/**/train_images.npy", recursive=True))[0])
code_src = os.path.dirname(sorted(glob.glob(inp + "/**/run_all.py", recursive=True))[0])
shutil.copytree(code_src, "/kaggle/working/code", dirs_exist_ok=True)
print("data:", data_dir, os.listdir(data_dir))
print("code:", code_src)
def run(args):
    p = subprocess.Popen([sys.executable] + args, cwd="/kaggle/working/code",
                         stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    for line in p.stdout:
        print(line, end="", flush=True)
    assert p.wait() == 0, "command failed"
subprocess.run(["nvidia-smi"])
prev = os.path.dirname(sorted(glob.glob(inp + "/**/phaseA_status.json", recursive=True))[0])
print("phase A outputs:", prev)
```

```python
run(["run_all.py", "--data", data_dir, "--out", "/kaggle/working/runs", "--phase", "B", "--prev", prev])
```

```python
run(["make_submission.py", "--data", data_dir, "--out", "/kaggle/working/runs", "--sub", "/kaggle/working/submission"])
import json
print(json.dumps(json.load(open("/kaggle/working/submission/results.json")), indent=1))
```

```python
for f in glob.glob("/kaggle/working/runs/*.zip"): os.remove(f)
shutil.make_archive("/kaggle/working/ALL_RESULTS", "zip", "/kaggle/working", "runs")
shutil.make_archive("/kaggle/working/SUBMISSION_FILES", "zip", "/kaggle/working", "submission")
print(os.path.getsize("/kaggle/working/ALL_RESULTS.zip")/1e6, "MB")
```
